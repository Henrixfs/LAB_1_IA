```python
def agente_riego(humedad_suelo, llovera_hoy):
    """
    Agente reactivo simple (dos percepciones combinadas, sin memoria).

    PEAS:
    - Percepcion (S): humedad_suelo (0-100) y llovera_hoy (True/False).
    - Accion (A): regar o no regar.
    - Entorno (E): jardin cuya humedad varia por riego, lluvia y evaporacion.
    - Objetivo: mantener humedad adecuada sin desperdiciar agua.
    - Medida de desempeño (P): evitar riego innecesario y evitar suelo muy seco.
    """
    umbral_seco = 30  # por debajo de esto se considera suelo seco

    if humedad_suelo < umbral_seco and not llovera_hoy:
        return "Regar"
    else:
        return "No regar"
```

```python
casos = [
    (10, False),  # muy seco, no lloverá -> debería regar
    (10, True),   # muy seco, PERO lloverá -> condiciones compiten
    (70, False),  # húmedo, no lloverá -> no necesita
    (70, True),   # húmedo y lloverá -> no necesita
    (25, False),  # seco, no lloverá -> debería regar
    (50, True),   # ni seco ni lloverá relevante -> no regar
]

for humedad_suelo, llovera_hoy in casos:
    accion = agente_riego(humedad_suelo, llovera_hoy)
    print(f"Humedad: {humedad_suelo} | Lloverá hoy: {llovera_hoy} -> Accion: {accion}")
```
