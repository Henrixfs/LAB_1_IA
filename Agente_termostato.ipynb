```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22, margen=1.0):
    """
    Agente reactivo simple (sin memoria, sin modelo del entorno).

    PEAS:
    - Percepcion (S): temperatura actual del ambiente (C).
    - Accion (A): encender aire acondicionado, encender calefaccion, o no hacer nada.
    - Entorno (E): ambiente cerrado cuya temperatura varia por factores externos.
    - Objetivo: mantener la temperatura cerca de la temperatura_objetivo.
    - Medida de desempeño (P): tiempo que la temperatura permanece dentro del margen.
    """
    diferencia = temperatura_actual - temperatura_objetivo
    if diferencia > margen:
        return "Encender aire acondicionado (enfriar)"
    elif diferencia < -margen:
        return "Encender calefaccion (calentar)"
    else:
        return "No hacer nada (temperatura OK)"
```

```python
percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5, 21.0, 27.2]

for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepcion: {t} C -> Accion del agente: {accion}")
```
